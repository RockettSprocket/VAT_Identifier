```python
import polars as pl

path = r"D:\VAT Identifier Discovery\runs\run_02_directory_filter_fallback_search\pipeline_results.csv"
df = pl.read_csv(path, infer_schema=False)  # all text, so company numbers keep leading zeros

df.group_by("outcome").agg(pl.len().alias("count")).sort("count", descending=True)
```

```python
vat = (
    df.filter(pl.col("vat_number").str.strip_chars() != "")
      .select(["CompanyNumber", "CompanyName", "vat_number"])
)

vat.write_csv(r"D:\VAT Identifier Discovery\runs\run_02_directory_filter_fallback_search\vat_found.csv")
vat
```
