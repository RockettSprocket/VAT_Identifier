```python
from pathlib import Path

sizes = {"likely": 60, "unclear": 200, "dormant": 40,
         "new_no_accounts": 40, "overdue_no_accounts": 40, "subsidiary": 40}

manual_cols = [
    "outcome",                 # no_website_found | website_no_vat | vat_found
    "search_query", "website", "site_match_evidence", "company_number_on_site",
    "vat_number", "vat_location_on_site",
    "hmrc_valid",              # yes | no
    "hmrc_name", "hmrc_address",
    "match_verdict",           # same | different | unsure
    "notes",
]

sample = (
    pop.filter(pl.col("vat_bucket").is_in(list(sizes)))
       .filter(pl.int_range(pl.len()).shuffle(seed=42).over("vat_bucket")
               < pl.col("vat_bucket").replace_strict(sizes))
       .with_columns(*[pl.lit(None, dtype=pl.String).alias(c) for c in manual_cols])
       .sample(fraction=1.0, shuffle=True, seed=42)
)

# Generated file: safe to overwrite on every run
sample.write_csv("data/sample_blank.csv")

# Working file: created once, never overwritten by code
work = Path("data/sample_work.csv")
if not work.exists():
    sample.write_csv(work)
else:
    print("sample_work.csv already exists - not touching it")
```

```python
# Results - run whenever you want to see progress
checked = (pl.read_csv("data/sample_work.csv", infer_schema=False)
             .filter(pl.col("outcome").is_not_null()))

N = (pop.filter(pl.col("vat_bucket").is_in(list(sizes)))
        ["vat_bucket"].value_counts().rename({"count": "N"})
        .with_columns(weight=pl.col("N") / pl.col("N").sum()))

by_bucket = (
    checked.group_by("vat_bucket")
           .agg(
               n=pl.len(),
               has_website=(pl.col("outcome") != "no_website_found").sum(),
               found=(pl.col("outcome") == "vat_found").sum(),
               valid=(pl.col("hmrc_valid") == "yes").fill_null(False).sum(),
               confirmed=(pl.col("match_verdict") == "same").fill_null(False).sum(),
           )
           .with_columns(
               found_rate=pl.col("found") / pl.col("n"),
               confirmed_rate=pl.col("confirmed") / pl.col("n"),
               false_positive_rate=1 - pl.col("confirmed") / pl.col("found"),
           )
           .join(N, on="vat_bucket", how="right")
)
by_bucket
```

```python
# Weighted overall rate - only meaningful once every bucket has checked rows
by_bucket.select(
    (pl.col("found_rate") * pl.col("weight")).sum().alias("weighted_found_rate"),
    (pl.col("confirmed_rate") * pl.col("weight")).sum().alias("weighted_confirmed_rate"),
)
```
